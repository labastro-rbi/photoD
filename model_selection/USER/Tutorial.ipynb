# Tutorial for the inference of the trained models

## Pre-requisites
This is the tutorial on how to inference the trained models for the prediction of the stellar parameters. 
Pre-requisites to access the trained models are to install Tensorflow. In order to install Tensorflow please run the following command in the terminal.
```bash
pip install tensorflow
```
The trained models are saved in the folder 
```bash
USER/models/
```

## Importing the libraries

````python
import numpy as np
import tensorflow as tf
````

## Loading the trained models

````python
model = tf.keras.models.load_model('models/NaiveSingle.keras')
model.summary()
````

Model has two inputs and three outputs. The inputs are the values for the color filters and the corresponding errors. The outputs are the apsolute magnitude, extinction and metallicity.

## Preparing the inputs
Inputs should be a list of two numpy arrays in the shape
```python
input_shape = [(number_of_examples, 5), (number_of_examples, 5)]
```

````python
# inputs
number_of_examples = 1000000
colors = np.random.rand(number_of_examples, 5) # r_mag, u-g, g-r, r-i, i-z
colors_err = np.random.rand(number_of_examples, 5) # r_mag error, u-g error, g-r error, r-i error, i-z error
x = [colors, colors_err]
````

## Inference

````python
y = model(x)
````

## Interpreting the outputs
Outputs are a list of three numpy arrays in the shape
```python
output_shape = [(number_of_examples, 2), (number_of_examples, 2), (number_of_examples, 2)]
```
The first element of the list is the absolute magnitude and its error, the second element is the extinction and its error and the third element is the metallicity and its error.

````python
# outputs
Mr = y[0][:,0].numpy()
Mr_err = y[0][:,1].numpy()
A_r = y[1][:,0].numpy()
A_r_err = y[1][:,1].numpy()
FeH = y[2][:,0].numpy()
FeH_err = y[2][:,1].numpy()
print (Mr.shape, Mr_err.shape, A_r.shape, A_r_err.shape, FeH.shape, FeH_err.shape)
````
